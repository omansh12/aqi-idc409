```python
import matplotlib.pyplot as plt

# 1. Define data points (X and Y coordinates)
x_days = [1, 2, 3, 4, 5, 6, 7]
y_temp = [72, 74, 71, 75, 79, 80, 78]

# 2. Generate the line graph
plt.plot(x_days, y_temp, color='blue', marker='o', linestyle='-')

# 3. Add labels and a title to make it readable
plt.xlabel('Day of the Week')
plt.ylabel('Temperature (°F)')
plt.title('Weekly Temperature Trends')

# 4. Display the chart on your screen
plt.show()

```
